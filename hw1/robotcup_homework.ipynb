第一题

```python
a=[1,2,3,4]
num=0
flag=0
for i in a:
    for j in a:
        for k in a:
            num=num+1
            if i==j or i==k or j==k:
                flag=flag+1
s=num-flag
print(s)
```

第三题

```python
def fib(n):
    if n==1:
        return 0
    elif n==2 or n==3:
        return 1
    return fib(n-1)+fib(n-2)

for i in range(1,21):
    print(fib(i))
```
